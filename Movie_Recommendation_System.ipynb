```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
movies = {
      "Movie": [
              "Avatar",
                      "Titanic",
                              "Avengers",
                                      "Iron Man",
                                              "Batman",
                                                      "Superman",
                                                              "Frozen",
                                                                      "Toy Story",
                                                                              "Joker",
                                                                                      "Spider-Man"
                                                                                          ],

                                                                                              "Genre": [
                                                                                                      "Action Adventure",
                                                                                                              "Romance Drama",
                                                                                                                      "Action Superhero",
                                                                                                                              "Action Superhero",
                                                                                                                                      "Action Crime",
                                                                                                                                              "Action Hero",
                                                                                                                                                      "Animation Family",
                                                                                                                                                              "Animation Comedy",
                                                                                                                                                                      "Crime Drama",
                                                                                                                                                                              "Action Superhero"
                                                                                                                                                                                  ]
                                                                                                                                                                                  }

df = pd.DataFrame(movies)

print(df)
```

```python
cv = CountVectorizer()

matrix = cv.fit_transform(df["Genre"])

similarity = cosine_similarity(matrix)

print("Similarity Matrix Created")
```

```python
def recommend(movie):
    index = df[df["Movie"] == movie].index[0]
    distance = list(enumerate(similarity[index]))
    distance = sorted(distance, key=lambda x: x[1], reverse=True)
    print("Recommended Movies:\n")
    for i in distance[1:6]:
        print(df.iloc[i[0]].Movie)
```

```python
recommend("Avatar")
```

```python
recommend("Batman")
```
