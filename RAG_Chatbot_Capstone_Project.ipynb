```python
import os
import faiss
import numpy as np
import gradio as gr

from sentence_transformers import SentenceTransformer
from transformers import pipeline

from pypdf import PdfReader
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("lifebricksglobal/llm-rag-chatbot-training-dataset")

print("Path to dataset files:", path)
```

```python
import os

# Assuming the PDF is directly in the downloaded dataset path or a subdirectory
# You might need to adjust the filename if it's different or in a subfolder
# Let's list the files in the downloaded path to find the PDF.

pdf_files = [f for f in os.listdir(path) if f.endswith('.pdf')]

if not pdf_files:
    print(f"No PDF files found in the downloaded path: {path}")
    # Handle case where no PDF is found, or raise an error.
    # For now, we'll assume there's at least one and pick the first one.
    # You might need more sophisticated logic if there are multiple PDFs.
else:
    pdf_name = os.path.join(path, pdf_files[0])

    reader = PdfReader(pdf_name)

    text = ""

    for page in reader.pages:
        text += page.extract_text()

    print(text[:1000])
```

```python
chunk_size = 300

documents = [
    text[i:i+chunk_size]
        for i in range(0, len(text), chunk_size)
        ]

print("Chunks:", len(documents))
```

```python
embedder = SentenceTransformer("all-MiniLM-L6-v2")

embeddings = embedder.encode(documents)

embeddings = np.array(embeddings).astype("float32")

print(embeddings.shape)
```

```python
index = faiss.IndexFlatL2(embeddings.shape[1])

index.add(embeddings)

print("Documents Indexed Successfully")
```

```python
generator = pipeline(
      "text-generation",
          model="google/flan-t5-small"
          )
```

```python
def retrieve(query):

      query_embedding = embedder.encode([query])

      query_embedding = np.array(query_embedding).astype("float32")

      distance, index_result = index.search(query_embedding, 1)

      return documents[index_result[0][0]]
```

```python
def chatbot(question):

      context = retrieve(question)

      prompt = f"""
      Context:
      {context}

      Question:
      {question}

      Answer:
      """

      result = generator(
              prompt,
              max_length=100,
              do_sample=False
      )

      return result[0]["generated_text"]
```

```python
question = "What is Artificial Intelligence?"

answer = chatbot(question)

print("Question:", question)

print("Answer:", answer)
```

```python
question = "What is Artificial Intelligence?"

answer = chatbot(question)

print("Question:", question)

print("Answer:", answer)
```

```python
demo = gr.Interface(
      fn=chatbot,
      inputs="text",
      outputs="text",
      title="RAG Chatbot"
)

demo.launch()
```
